```python
XOR_INPUTS = ((0,0), (0,1), (1,0), (1,1),)
XOR_OUTPUTS = []

def xor_func(x1, x2):
    _x1, _x2 = x1, x2
    return _x1 ^ _x2

for x1, x2 in XOR_INPUTS:
    XOR_OUTPUTS.append(xor_func(x1, x2))

print(XOR_OUTPUTS)
```

```python
data = {'x1': [], 'x2': [], 'x1 XOR x2': []}
XOR_OUTPUTS = []

for x1, x2 in XOR_INPUTS:
    xor_result = xor_func(x1, x2)
    XOR_OUTPUTS.append(xor_result)
    data['x1'].append(x1)
    data['x2'].append(x2)
    data['x1 XOR x2'].append(xor_result)

df = pd.DataFrame(data)
df.style

```
