```python
XOR_INPUTS = ((0,0), (0,1), (1,0), (1,1),)
XOR_OUTPUTS = []

def xor_func(x1, x2):
    _x1, _x2 = x1, x2
    return _x1 ^ _x2

for x1, x2 in XOR_INPUTS:
    XOR_OUTPUTS.append(xor_func(x1, x2))

print(XOR_OUTPUTS)
```

```python
data = {'x1': [], 'x2': [], 'x1 XOR x2': []}
XOR_OUTPUTS = []

for x1, x2 in XOR_INPUTS:
    xor_result = xor_func(x1, x2)
    XOR_OUTPUTS.append(xor_result)
    data['x1'].append(x1)
    data['x2'].append(x2)
    data['x1 XOR x2'].append(xor_result)

df = pd.DataFrame(data)
df.style

```

```python
import itertools

params = {
    "w1": (0,1),
    "w2": (1,0), 
    "w3": (1,1),
    "w4": (0,1),
    "w5": (0,1),
    "w6": (0,1)
}

keys = list(params.keys())

combinations = list(itertools.product([-1 ,0, 1], repeat=len(keys)))

result = [dict(zip(keys, combo)) for combo in combinations]

for i, item in enumerate(result):
    print(f"{i+1}: {item}")
```

```python
params = {"w1":(0,1),"w2":(1,0),"w3":(1,1),}
th = 3

def neuron(x1: int, x2: int, inp_1: int, inp_2: int, th: int) -> int: 
    return x1 * inp_1 + x2 * inp_2 >= th

def check_xor(params: dict, th: int):
    answers = []
    for x1, x2 in XOR_INPUTS:
        w1_1 = params["w1"][0]
        w1_2 = params["w1"][1]
        w2_1 = params["w2"][0]
        w2_2 = params["w2"][1]
        w3_1 = params["w3"][0]
        w3_2 = params["w3"][1]

        neuron_1 = neuron(x1=x1, x2=x2, inp_1=w1_1, inp_2=w1_2, th=th)
        neuron_2 = neuron(x1=x1, x2=x2, inp_1=w2_1, inp_2=w2_2, th=th)
        neuron_3 = neuron(x1=neuron_1, x2=neuron_2, inp_1=w3_1, inp_2=w3_2, th=th)

        answers.append(neuron_3)
    return answers
```

```python
for theta in range(-2,2):
    for i, item in enumerate(result):
        answers = []
        for x1, x2 in XOR_INPUTS:
            w1_1 = item["w1"]
            w1_2 = item["w2"]
            w2_1 = item["w3"]
            w2_2 = item["w4"]
            w3_1 = item["w5"]
            w3_2 = item["w6"]

            neuron_1 = neuron(x1=x1, x2=x2, inp_1=w1_1, inp_2=w1_2, th=theta)
            neuron_2 = neuron(x1=x1, x2=x2, inp_1=w2_1, inp_2=w2_2, th=theta)
            neuron_3 = neuron(x1=neuron_1, x2=neuron_2, inp_1=w3_1, inp_2=w3_2, th=theta)

            answers.append(neuron_3)

        if answers == [False, True, True, False]:
            solution_data = {
                'Нейрон': ['1', '2', '3'],
                'Вес 1': [item['w1'], item['w3'], item['w5']],
                'Вес 2': [item['w2'], item['w4'], item['w6']],
                'Порог': [theta, theta, theta]
            }
            solution_df = pd.DataFrame(solution_data)

            print(solution_df.to_string(index=False))
```
